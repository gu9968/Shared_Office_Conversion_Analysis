# 00. Project Overview — 공유오피스 무료체험 결제 전환 예측

공유오피스 **3일 무료체험 고객의 이용 행동**을 분석하고,
결제 전환 가능성이 높은 고객을 사전에 식별하는 머신러닝 프로젝트입니다.

## 프로젝트 흐름

`원천 데이터 → 전처리 → EDA/행동변수 → 모델 비교 → XGBoost 튜닝 → Streamlit/FastAPI/n8n`

## 분석 대상

- 무료체험 신청자: 9,624명
- 실제 방문 고객: 6,132명
- 최종 머신러닝 표본: 5,629명
- 최종 모델: XGBoost Classifier
- 최종 입력 피처: 42개
- 운영 Threshold: 0.35

## 데이터 공개 정책

원본 데이터에는 사용자 식별키 및 이용 로그가 포함되어 있어 **GitHub 저장소에 업로드하지 않습니다.**
노트북 실행 시 본인 로컬 환경의 `data/raw/` 폴더에 원본 CSV 5개가 필요합니다.

```text
data/
└── raw/
    ├── site_area.csv
    ├── trial_register.csv
    ├── trial_visit_info.csv
    ├── trial_access_log.csv
    └── trial_payment.csv
```

저장소에서는 `data/raw/`, `data/interim/`, `data/processed/`를 Git 추적 대상에서 제외합니다.

## Notebook 구성

| Notebook | 역할 |
|---|---|
| `01_data_preprocessing.ipynb` | 원본 5종 정제 및 사용자 단위 기본 테이블 생성 |
| `02_eda_analysis.ipynb` | 행동 변수 생성 및 주요 전환 패턴 분석 |
| `03_model_experiment.ipynb` | Logistic / Random Forest / XGBoost 비교 |
| `04_model_tuning_final.ipynb` | 42개 피처 생성, 최종 XGBoost 학습 및 모델 파일 저장 |
| `05_streamlit_prepare.ipynb` | 저장된 모델 자산 검증 및 예측 샘플 생성 |
